Import Libraries

```python
from dotenv import load_dotenv
import os
import pandas as pd
from pandasai import SmartDataframe
from pandasai.llm import OpenAI
```

Initialising Smart Dataframe

```python
load_dotenv()
openai_key=os.getenv("OPENAI_API_KEY")
llm=OpenAI(api_token=openai_key)
df=pd.read_csv("data/nasa_tweets.csv")
smart_df=SmartDataframe(df, config={"llm": llm})
```

Chat with the CSV file

```python
smart_df.chat("Average amount of likes for all tweets containing text")
```
